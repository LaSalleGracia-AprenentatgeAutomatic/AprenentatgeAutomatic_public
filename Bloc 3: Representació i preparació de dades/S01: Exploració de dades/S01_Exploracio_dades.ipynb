
## 📊 Estadística descriptiva bàsica
Calcula mesures centrals i de dispersió de les variables numèriques (p. ex., `math score`, `reading score`, `writing score`).

> **Recordatori matemàtic (Just In Time):**
> - Mitjana: $\bar{x} = \frac{1}{n}\sum x_i$
> - Desviació estàndard: $\sigma = \sqrt{\frac{1}{n}\sum (x_i - \bar{x})^2}$
> - Mediana: valor central de les dades ordenades.


```python

# TODO: Substitueix per les columnes numèriques del teu dataset si tenen un nom diferent
num_cols = ["math score", "reading score", "writing score"]
stats = pd.DataFrame({
    "mitjana": [data[c].mean() for c in num_cols],
    "mediana": [data[c].median() for c in num_cols],
    "desviacio_estandard": [data[c].std() for c in num_cols],
}, index=num_cols)
stats
```
